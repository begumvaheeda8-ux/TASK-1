```python
# Project 1: Rule-Based AI Chatbot
# Artificial Intelligence
# Developed using Python

print("Rule-Based AI Chatbot")
print("Welcome! I am your simple AI chatbot.")
print("Type 'bye' to exit.")

responses = {
    "hi": "Hello! How can I help you?",
    "hello": "Hello! How can I help you?",
    "hey": "Hi there!",
    "how are you": "I am doing great! Thank you for asking.",
    "what is your name": "I am a Rule-Based AI Chatbot.",
    "what can you do": "I can respond to simple predefined questions.",
    "help": "You can greet me, ask my name, ask how I am, or type bye to exit."
}

while True:
    user_input = input("You: ").lower().strip()

    if user_input in ["bye", "exit", "quit"]:
        print("Bot: Goodbye! Have a nice day!")
        break

    response = responses.get(
        user_input,
        "Sorry, I don't understand that."
    )

    print("Bot:", response)
```
